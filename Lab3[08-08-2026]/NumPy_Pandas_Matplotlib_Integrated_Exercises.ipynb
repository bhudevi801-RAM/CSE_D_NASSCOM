## Synthetic Data Generation (used throughout this notebook)

**What it is:** synthetic data is data you *generate with code* instead of collecting from the real world — fake-but-realistic numbers that follow a distribution you choose on purpose. Every exercise below starts with a line like `rng.normal(...)` or `rng.choice(...)` to create a dataset out of thin air before running it through Pandas and Matplotlib.

**Why we do it (instead of using a real CSV):**
1. **No dataset needed to practice.** You can invent a dataset with any shape, size, or relationship you want, on demand.
2. **You control the "ground truth."** Because *you* decide the mean/spread/relationship when generating the data, you can check whether your analysis correctly *recovers* that truth (e.g. "I generated this with mean 70 — does `df['score'].mean()` come out close to 70?").
3. **Reproducibility.** Real-world randomness can't be replayed. Code-generated randomness can, if you fix the **seed**.

**The tool: NumPy's `Generator` object (`np.random.default_rng`)**

```python
rng = np.random.default_rng(42)   # 42 is the "seed" — any fixed integer works
```

A `Generator` produces pseudo-random numbers — they *look* random but are fully determined by the seed. Same seed → same numbers, every time you rerun the notebook. That's why every exercise starts from the same `rng` created once at the top of this notebook, instead of calling `np.random.default_rng()` fresh (no seed) inside each exercise.

**The distributions used in this notebook:**

| Call | Produces | Used for |
|---|---|---|
| `rng.normal(loc, scale, size)` | Bell-curve-shaped data around a mean (`loc`) with a spread (`scale`) | Exam scores, heights, salaries — most real measurements cluster this way |
| `rng.uniform(low, high, size)` | Every value in `[low, high)` equally likely | Study hours, arrival times — things with no natural "typical" value |
| `rng.choice(options, size)` | Randomly picks from a fixed list of categories | Departments, colors, class labels |
| `rng.integers(low, high, size)` | Random whole numbers in `[low, high)` | Ages, counts, discrete quantities |

