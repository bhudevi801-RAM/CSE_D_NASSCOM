## Section 2 — Vector Operations

**Math idea:** the **dot product** `v · w = Σ vᵢwᵢ` tells you how much two vectors point in the same direction. The **norm** `‖v‖ = √(Σvᵢ²)` tells you a vector's length.

````python
# WORKED EXAMPLE
v = np.array([1, 2, 3])
w = np.array([4, 5, 6])

print("v + w      =", v + w)
print("2 * v      =", 2 * v)
print("v . w      =", np.dot(v, w))
print("||v||      =", np.linalg.norm(v))
print("distance(v,w) =", np.linalg.norm(v - w))
````

## Section 3 — Matrix Operations

**Math idea:** `*` multiplies element-by-element. `@` performs true **matrix multiplication**: `(AB)ᵢⱼ = Σₖ Aᵢₖ Bₖⱼ`. These are NOT the same thing — this is the #1 beginner mistake in NumPy.

````python
# WORKED EXAMPLE
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A * B (element-wise):\n", A * B)
print("A @ B (matrix multiplication):\n", A @ B)
print("A transpose:\n", A.T)
print("determinant of A:", np.linalg.det(A))
print("inverse of A:\n", np.linalg.inv(A))

# Solve Ax = b
b = np.array([9, 8])
x = np.linalg.solve(A, b)
print("solution x of A x = b:", x)
````

## Section 4 — Broadcasting

**Math idea:** broadcasting lets NumPy combine arrays of different shapes by conceptually "stretching" the smaller one. This is how we mean-center data: subtract a single row (the column means) from every row of a matrix.

````python
# WORKED EXAMPLE
data = np.array([
    [10, 200],
    [12, 220],
    [ 9, 180],
    [14, 240],
])

col_means = data.mean(axis=0)
print("column means:", col_means)

centered = data - col_means   # broadcasting: (4,2) - (2,) -> (4,2)
print("centered data:\n", centered)
print("new column means (should be ~0):", centered.mean(axis=0))
````

## Section 5 — Basic Statistics

**Math idea:** mean `μ = (1/n)Σxᵢ`, variance `σ² = (1/n)Σ(xᵢ-μ)²`, standard deviation `σ = √σ²`. These describe the center and spread of a dataset.

````python
# WORKED EXAMPLE
scores = np.array([55, 60, 65, 70, 70, 75, 80, 95])

print("mean:", scores.mean())
print("variance:", scores.var())
print("std:", scores.std())

z_scores = (scores - scores.mean()) / scores.std()
print("z-scores:", z_scores)
````

## Section 6 — Eigenvalues, Eigenvectors & SVD (preview)

**Math idea:** `A v = λ v` — an eigenvector `v` of matrix `A` keeps its direction under the transformation `A`, only getting scaled by the eigenvalue `λ`. We'll use this heavily when we cover PCA in the scikit-learn module; for now, just get comfortable calling the functions and reading the output.

````python
# WORKED EXAMPLE
A = np.array([[2, 0], [0, 3]])
eigenvalues, eigenvectors = np.linalg.eig(A)
print("eigenvalues:", eigenvalues)
print("eigenvectors:\n", eigenvectors)

# Verify: A @ v should equal lambda * v for each eigenvector
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    print(f"A @ v = {A @ v},  lambda * v = {lam * v}")
````
