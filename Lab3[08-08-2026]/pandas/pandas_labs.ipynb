## Section 2 — Filtering ↔ Set Theory

**Math idea:** `{x ∈ S | P(x)}` — a boolean mask is the indicator function of a predicate `P`. `&` is set intersection (AND), `|` is set union (OR).

````python
# WORKED EXAMPLE
mask = df["math"] > 75
print(mask)             # the indicator function itself: True/False per row

high_math = df[mask]
print(high_math)

strong_students = df[(df["math"] > 75) & (df["science"] > 75)]
print(strong_students)
````

## Section 3 — GroupBy ↔ Set Partitions + Aggregation

**Math idea:** `groupby` partitions rows into non-overlapping subsets by key, then applies a `Σ`-style aggregation to each subset independently.

````python
# WORKED EXAMPLE
sales = pd.DataFrame({
    "region": ["North", "North", "South", "South", "East"],
    "amount": [100, 150, 200, 50, 300],
})

grouped_sum = sales.groupby("region")["amount"].sum()
print(grouped_sum)

summary = sales.groupby("region")["amount"].agg(["sum", "mean", "count"])
print(summary)
````

## Section 4 — Merge / Join ↔ Relational Algebra

**Math idea:** `inner` join = intersection of keys, `outer` join = union of keys, `left`/`right` keep all of one side.

````python
# WORKED EXAMPLE
students = pd.DataFrame({
    "student_id": [1, 2, 3],
    "name": ["Asha", "Ravi", "Meera"],
})
grades = pd.DataFrame({
    "student_id": [1, 2, 4],
    "grade": ["A", "B", "A"],
})

print("inner (intersection):\n", pd.merge(students, grades, on="student_id", how="inner"))
print("left:\n", pd.merge(students, grades, on="student_id", how="left"))
print("outer (union):\n", pd.merge(students, grades, on="student_id", how="outer"))
````

## Section 5 — Descriptive Stats & Covariance/Correlation Matrix

**Math idea:** column-wise mean/std are the same formulas from the NumPy module. The **covariance matrix** shows how every pair of columns varies together — and its eigenvectors (via `np.linalg.eig`) are exactly what PCA uses.

````python
# WORKED EXAMPLE
print(df.mean())
print(df.std())
print(df.describe())

cov_matrix = df.cov()
corr_matrix = df.corr()
print("covariance matrix:\n", cov_matrix)
print("correlation matrix:\n", corr_matrix)
````

## Section 6 — Pivot Tables ↔ Cross-Tabulation

**Math idea:** turning long-format data into a matrix where rows/columns are two categorical variables — a reshape, like NumPy's `.reshape()` but keyed by labels instead of position.

````python
# WORKED EXAMPLE
records = pd.DataFrame({
    "student": ["Asha", "Asha", "Ravi", "Ravi"],
    "subject": ["math", "science", "math", "science"],
    "score":   [85, 90, 70, 65],
})

pivoted = records.pivot(index="student", columns="subject", values="score")
print(pivoted)
````

## Section 7 — Missing Data ↔ Partial Functions & Three-Valued Logic

**Math idea:** a column is a function from index to value. Missing data makes it a **partial** function — defined on part of the domain, undefined elsewhere. `isnull()` is the domain-membership predicate (same indicator-function idea as Section 2); `dropna()` restricts to where the function is defined; `fillna()` extends it back to a total function.

````python
# WORKED EXAMPLE
df_missing = df.copy()
df_missing.loc["Ravi", "math"] = np.nan
df_missing.loc["Meera", "science"] = np.nan
print(df_missing)

print(np.nan == np.nan)        # False -- "unknown = unknown" is still unknown
print(df_missing.isnull())     # indicator function, same idea as Section 2's boolean mask
print(df_missing.isnull().sum())
````
