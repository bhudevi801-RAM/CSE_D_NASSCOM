*(see NumPy_Definitions.md → Why NumPy Arrays for the full explanation)*

```python
# ------------------------------------------------------------- #
# 🔹 1. WHY NUMPY — DEMO
# ------------------------------------------------------------- #

python_list = list(range(1_000_000))
numpy_array = np.arange(1_000_000)

import time

start = time.time()
doubled_list = [x * 2 for x in python_list]
list_time = time.time() - start

start = time.time()
doubled_array = numpy_array * 2
array_time = time.time() - start

print(f"Python list loop: {list_time:.4f} sec")
print(f"NumPy vectorized: {array_time:.4f} sec")
print(f"NumPy was ~{list_time / array_time:.0f}x faster")
```
