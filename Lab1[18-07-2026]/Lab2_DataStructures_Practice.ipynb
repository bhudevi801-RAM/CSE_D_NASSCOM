### 2. Tuples — Fixed Exam Slot
Create a tuple `exam_slot = ("Computer Networks", "2:00 PM")`. Unpack it into `subject` and `time` variables and print them. Then try to change the time (`exam_slot[1] = "3:00 PM"`) inside a `try/except` block and print the error message you get — this proves tuples are immutable.

```python
# TODO: 2. Tuples solution
# 2. Tuples solution: Verifying immutability of administrative slots
exam_slot = ("Computer Networks", "2:00 PM")

# Unpack tuple into variables
subject, time = exam_slot
print(f"Unpacked Exam Schedule -> Subject: {subject}, Time: {time}")

# Attempting mutation within a protective block
try:
    exam_slot[1] = "3:00 PM"
except TypeError as error:
    print(f"Caught Immutability Error: {error}")
```

### 5. Sets — Comparing Two Students' Skills
Create two sets, `your_skills` and `friend_skills`, each with at least 4 programming languages/tools (with some overlap). Print:
- Skills you **both** know (intersection)
- All skills **across both of you** (union)
- Skills only **you** know (difference)

```python
# TODO: 5. Sets solution
# 5. Sets solution: Venn operations on tech stacks
your_skills = {"Python", "Java", "C++", "SQL", "Git"}
friend_skills = {"JavaScript", "Python", "HTML", "SQL", "Docker"}

# 1. Intersection (Skills both know)
both_know = your_skills.intersection(friend_skills)
print(f"Shared Tech Stack: {both_know}")

# 2. Union (All unique skills across both profiles)
all_skills = your_skills.union(friend_skills)
print(f"Combined Skill Pool: {all_skills}")

# 3. Difference (Skills only you know)
only_you = your_skills.difference(friend_skills)
print(f"Your Unique Advantage: {only_you}")
```

### 6. Dictionaries — Nested Gradebook

## Combined Task — Semester Report (submit this)

Bring everything together into one program that:
1. Defines a nested `gradebook` **dictionary** (at least 6 subjects, with `marks` and `credits`).
2. Defines a `lab_submissions` **list** with at least 10 entries (with repeats).
3. Computes the **unique subjects submitted** using a **set**.
4. Computes the **high-scoring subjects (marks >= 85)** using a **set comprehension**.
5. Computes the **best subject** as a **tuple** `(subject_name, marks)` — the highest-scoring subject overall.
6. Prints all of the above in a clearly labelled semester report.


```python
# TODO: Combined task solution
# 1. Definitions (Nested Gradebook with 6 subjects)
semester_gradebook = {
    "OS": {"marks": 88, "credits": 4, "faculty": "Dr. Ramesh"},
    "DBMS": {"marks": 92, "credits": 4, "faculty": "Prof. Priya"},
    "Networks": {"marks": 78, "credits": 3, "faculty": "Dr. Amit"},
    "Compiler Design": {"marks": 85, "credits": 4, "faculty": "Prof. Sunil"},
    "Web Tech": {"marks": 95, "credits": 3, "faculty": "Dr. Kavitha"},
    "Data Structures": {"marks": 82, "credits": 4, "faculty": "Prof. Anjali"}
}

# 2. Defininig a lab_submissions list with 10 entries (with repeats)
weekly_lab_submissions = ["OS", "DBMS", "OS", "Networks", "Compiler Design", "OS", "DBMS", "Web Tech", "Data Structures", "Networks"]

# 3. Compute unique subjects submitted using a set
unique_labs = set(weekly_lab_submissions)

# 4. Compute high-scoring subjects (marks >= 85) using a set comprehension
# (Ensures it filters from subjects actually submitted in weekly_lab_submissions)
high_scoring_labs = {subject for subject in unique_labs if subject in semester_gradebook and semester_gradebook[subject]["marks"] >= 85}

# 5. Compute the best subject as a tuple (subject_name, marks)
# Uses loop tracking to calculate the true maximum value safely without dependencies
best_subject_name = None
max_score = -1

for current_subject, details in semester_gradebook.items():
    if details["marks"] > max_score:
        max_score = details["marks"]
        best_subject_name = current_subject

best_subject_record = (best_subject_name, max_score)

# 6. Printing the clearly labelled Semester Report
print("=" * 60)
print("             CSE STUDENT SEMESTER PERFORMANCE REPORT         ")
print("=" * 60)
print(f"Total Lab Uploads Processed : {len(weekly_lab_submissions)}")
print(f"Unique Subjects Completed   : {unique_labs}")
print(f"Distinct Subject Count      : {len(unique_labs)}")
print("-" * 60)
print(f"High-Scoring Labs (>=85)    : {high_scoring_labs}")
print(f"Top Academic Achievement    : Subject: '{best_subject_record[0]}' | Marks: {best_subject_record[1]}")
print("-" * 60)

print("Detailed Grade Breakdown:")
for subject, data in semester_gradebook.items():
    status = "🌟 Excellent" if data["marks"] >= 85 else "✅ Cleared"
    print(f" - {subject:<16} | Marks: {data['marks']} | Credits: {data['credits']} | Instructor: {data['faculty']} [{status}]")
print("=" * 60)


```
