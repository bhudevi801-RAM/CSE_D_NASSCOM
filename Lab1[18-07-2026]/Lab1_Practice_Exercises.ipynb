### B8. Functions — `calculate_bill()`
Write a function `calculate_bill(cart, menu, is_parcel=False, gst_rate=0.05)` that applies the discount logic from B3 and returns a dictionary with `subtotal`, `discount`, `gst`, `packing_charge`, and `total`. Call it for two different carts to prove it's reusable.

````python
# B8 solution: Encapsulated business calculations
def calculate_bill(order_cart, business_menu, is_parcel=False, gst_rate=0.05):
    # Compute base values
    subtotal_val = 0.0
    for item in order_cart:
        subtotal_val += business_menu[item]
        
    # Apply discount bands
    if subtotal_val > 200:
        discount = subtotal_val * 0.10
    elif subtotal_val > 100:
        discount = subtotal_val * 0.05
    else:
        discount = 0.0
        
    gst = subtotal_val * gst_rate
    packing_charge = 5.0 if is_parcel else 0.0
    total_val = subtotal_val - discount + gst + packing_charge
    
    return {
        "subtotal": subtotal_val,
        "discount": discount,
        "gst": gst,
        "packing_charge": packing_charge,
        "total": total_val
    }

# Execution test with two independent carts
cart_one = ["Samosa", "Cold Coffee", "Vada"]  # 15 + 30 + 12 = 57
cart_two = ["Cold Coffee", "Cold Coffee", "Cold Coffee", "Cold Coffee", "Cold Coffee", "Cold Coffee", "Cold Coffee"] # 210

print("Cart 1 Bill:", calculate_bill(cart_one, menu, is_parcel=True))
print("Cart 2 Bill:", calculate_bill(cart_two, menu, is_parcel=False))
````

### B10. Exception Handling — Safer Billing Counter
Write `bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False)` that: (a) raises/catches a `KeyError` if an item in the cart isn't in `menu`, (b) raises/catches a `ValueError` if `cash_paid` is less than the bill total, and (c) otherwise prints the change due. Test all three cases.

````python
# B10 solution: Input sanitation and structural assertions
def bill_and_collect_cash(order_cart, business_menu, cash_paid, is_parcel=False):
    try:
        # Assert item existence inside menu boundaries
        subtotal_val = 0.0
        for item in order_cart:
            if item not in business_menu:
                raise KeyError(f"Item '{item}' is missing from the active menu matrix.")
            subtotal_val += business_menu[item]
            
        bill_data = calculate_bill(order_cart, business_menu, is_parcel)
        required_amt = bill_data["total"]
        
        # Assert cash collection criteria
        if cash_paid < required_amt:
            raise ValueError(f"Insufficient cash! Tendered: Rs.{cash_paid:.2f}, Required: Rs.{required_amt:.2f}")
            
        change_due = cash_paid - required_amt
        print(f"Transaction Success! Change due: Rs.{change_due:.2f}")
        return bill_data
        
    except KeyError as ke:
        print(f"Error Caught [KeyError]: {ke}")
    except ValueError as ve:
        print(f"Error Caught [ValueError]: {ve}")

# Operational verification runs
print("--- Testing Exceptions ---")
# Case A: KeyError
bill_and_collect_cash(["Pizza"], menu, 100) 
# Case B: ValueError
bill_and_collect_cash(["Cold Coffee", "Samosa"], menu, 20) 
# Case C: Valid Transaction
bill_and_collect_cash(["Cold Coffee", "Samosa"], menu, 100) 
````

````python
# TODO: Part C solution


import os

# 1. Menu definitions (Minimum 6 items)
canteen_menu = {
    "Samosa": 15.0,
    "Vada": 12.0,
    "Idli": 20.0,
    "Dosa": 40.0,
    "Tea": 10.0,
    "Cold Coffee": 30.0
}

# 2. Billing Core
def calculate_bill_system(cart, menu_data, is_parcel=False):
    subtotal_val = 0.0
    for item in cart:
        subtotal_val += menu_data[item]
        
    if subtotal_val > 200:
        discount = subtotal_val * 0.10
    elif subtotal_val > 100:
        discount = subtotal_val * 0.05
    else:
        discount = 0.0
        
    gst = subtotal_val * 0.05
    packing_charge = 5.0 if is_parcel else 0.0
    total_val = subtotal_val - discount + gst + packing_charge
    
    return {
        "subtotal": subtotal_val,
        "discount": discount,
        "gst": gst,
        "packing_charge": packing_charge,
        "total": total_val
    }

# 3. Log Exporter
def log_transaction(customer_name, final_total):
    with open("my_canteen_sales_log.txt", "a") as file:
        file.write(f"{customer_name}: Rs.{final_total:.2f}\n")

# Clear files from previous mini-tasks to print pristine log balances
if os.path.exists("my_canteen_sales_log.txt"):
    os.remove("my_canteen_sales_log.txt")

# 4. Mock Sequential Orders Database
customer_queue = [
    {"name": "  ravi SHARMA ", "cart": ["Dosa", "Cold Coffee", "Samosa"], "parcel": True, "cash": 100.0}, # Valid
    {"name": "Meena", "cart": ["Dosa", "Burger"], "parcel": False, "cash": 150.0},                     # KeyError (Burger)
    {"name": "Arjun", "cart": ["Dosa", "Dosa", "Dosa", "Cold Coffee"], "parcel": False, "cash": 50.0}, # ValueError (Low Cash)
    {"name": "Kiran Pod", "cart": ["Idli", "Tea"], "parcel": False, "cash": 50.0}                     # Valid
]

print("--- Initializing Part C Order Processing Engine ---")

# 5. Core Operational Loop
for customer in customer_queue:
    clean_name = customer["name"].strip().title()
    print(f"\nProcessing Order for: {clean_name}")
    
    try:
        # Check Item validity
        for item in customer["cart"]:
            if item not in canteen_menu:
                raise KeyError(f"Item '{item}' not found on the canteen menu.")
                
        # Calculate pricing matrices
        invoice = calculate_bill_system(customer["cart"], canteen_menu, customer["parcel"])
        bill_total = invoice["total"]
        
        # Verify Payment settlement
        if customer["cash"] < bill_total:
            raise ValueError(f"Insufficient cash provided. Owed: Rs.{bill_total:.2f}, Paid: Rs.{customer['cash']:.2f}")
            
        # Complete settlement transactions
        change = customer["cash"] - bill_total
        print(f" -> Bill Total: Rs.{bill_total:.2f} | Tendered: Rs.{customer['cash']:.2f} | Change: Rs.{change:.2f}")
        
        log_transaction(clean_name, bill_total)
        print(" -> Transaction logged successfully.")
        
    except KeyError as error:
        print(f" -> [ABORTED] Menu Error: {error}")
    except ValueError as error:
        print(f" -> [ABORTED] Cash Error: {error}")

# 6. Read and verify finalized records
print("\n" + "="*40)
print("FINAL DAY LOG (my_canteen_sales_log.txt):")
print("="*40)
with open("my_canteen_sales_log.txt", "r") as file:
    print(file.read())
````
