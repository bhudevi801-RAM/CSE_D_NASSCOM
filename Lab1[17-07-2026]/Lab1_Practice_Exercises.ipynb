### B8. Functions — `calculate_bill()`
Write a function `calculate_bill(cart, menu, is_parcel=False, gst_rate=0.05)` that applies the discount logic from B3 and returns a dictionary with `subtotal`, `discount`, `gst`, `packing_charge`, and `total`. Call it for two different carts to prove it's reusable.

````python
# TODO: B8 solution
# write a function to calculate_bill(cart,menu_items,is_parcel=False, gst_rate=0.05) 
# apply the same logic as in B2 and B3 to calculate the final bill amount
#returns a dictionary with keys: subtotal, gst_amount, total_amount, discount_amount, packing_charge, final_amount
def calculate_bill(cart, menu_items, is_parcel=False, gst_rate=0.05):
    subtotal = sum(menu_items[item] * quantity for item, quantity in cart.items())
    gst_amount = subtotal * gst_rate
    total_amount = subtotal + gst_amount
    
    loyalty_discount = 0.10 if total_amount > 200 else 0
    discount_amount = total_amount * loyalty_discount
    packing_charge = 5.00 if is_parcel else 0.00
    final_amount = total_amount - discount_amount + packing_charge
    
    return {
        "subtotal": subtotal,
        "gst_amount": gst_amount,
        "total_amount": total_amount,
        "discount_amount": discount_amount,
        "packing_charge": packing_charge,
        "final_amount": final_amount
    }
````

### B10. Exception Handling — Safer Billing Counter
Write `bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False)` that: (a) raises/catches a `KeyError` if an item in the cart isn't in `menu`, (b) raises/catches a `ValueError` if `cash_paid` is less than the bill total, and (c) otherwise prints the change due. Test all three cases.

````python
# TODO: B10 solution
#write a bill_and_collect_cash(cart, menu_items, cash_paid, is_parcel=False)
#(a)raises/catches a keyerror if an item in the cart is not present in the menu_items dictionary
#(b)raises/catches a valueerror if cash_paid is less than the final_amount
#(c)prints out the change to be returned if cash_paid is greater than final_amount
def bill_and_collect_cash(cart, menu_items, cash_paid, is_parcel=False):
    try:
        bill = calculate_bill(cart, menu_items, is_parcel)
    except KeyError as e:
        print(f"Error: Item {e} is not available in the menu.")
        return

    if cash_paid < bill["final_amount"]:
        print("Error: Insufficient funds.")
        return

    change = cash_paid - bill["final_amount"]
    print(f"Change to be returned: ${change:.2f}")
````
